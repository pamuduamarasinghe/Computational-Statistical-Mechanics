1) Maxwell-Boltzmann Statistics – Paramagnetism

```python
import numpy as np
import matplotlib.pyplot as plt

eval = [-1,1]
N = 100

#constants
m=1
B=1
beta = 1
Emean = 0
z = 0

#partition function
for e in eval:
    z = z + np.exp(-beta * e)

Z = z**N

for e in eval :
    nmean = N*np.exp(-beta * e) / z
    Emean = Emean + nmean*e

print(f'Partition Funtion Z = {Z}')
print (f'Mean Numbers = {nmean}')
print (f'Mean Energy = {Emean}')


    



```
