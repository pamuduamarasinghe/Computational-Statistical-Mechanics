1) Maxwell-Boltzmann Statistics – Paramagnetism

```python
import numpy as np
import matplotlib.pyplot as plt

eval = [-1,1]
N = 100

#constants
m=1
B=1
beta = 1
Emean = 0
z = 0

#partition function
for e in eval:
    z = z + np.exp(-beta * e)

Z = z**N

for e in eval :
    nmean = N*np.exp(-beta * e) / z
    Emean = Emean + nmean*e

print(f'Partition Funtion Z = {Z}')
print (f'Mean Numbers = {nmean}')
print (f'Mean Energy = {Emean}')


    
```

```python
import numpy as np
import matplotlib.pyplot as plt

eval = [-1,1]
N = 100
#constants
m=1
B=1

beta = np.linspace(0.1 ,5,100)

n1 = []
n2 = []
z = []
Emean =[]

for b in beta :
    z = np.sum(np.exp(b*np.array(eval)))
    n_1 = (N*np.exp(-b * eval[0]) / z) 
    n_2 = (N*np.exp(-b * eval[1]) / z)
    Emean.append( n_1*eval[0]+ n_2*eval[1] )
    n1.append(n_1)
    n2.append(n_2)  

plt.figure(figsize=(10,5))
plt.plot(beta,n1, label='n1') 
plt.plot(beta,n2, label='n2')

plt.xlabel('Beta Values')
plt.ylabel('Mean number of particles')
plt.title('Mean number of particles in each energy state vs Beta') 
plt.legend()


#single particle as a function of beta

z1 =[]
for b in beta:
    z1.append(np.sum(np.exp(b*np.array(eval))))

plt.figure(figsize=(10,5))
plt.plot(beta,z1) 
plt.xlabel('Beta Values')
plt.ylabel('Partition function')
plt.title('Partition function of a single particle as a function of Beta') 


#internal energy of 1000 particles as a function of beta
plt.figure(figsize=(10, 6))
plt.plot(beta, Emean, label='Mean Energy')
plt.xlabel('Beta')
plt.ylabel('Mean Energy')
plt.title(f'Mean Energy vs Beta for N={N} particles')    
plt.legend()
plt.show()

```
