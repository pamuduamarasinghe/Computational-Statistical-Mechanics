Monte Carlo simulation of 1D Ising model, 
Average magnetization 〈𝑀〉

```python
import numpy as np

# Constants
N = 10
J = 1
H = 1
T = 1
kb = 1
steps = 100000

# Initial spins
spins = np.random.choice([-1, 1], size=N)

M_val = []

# Markov Ising simulation
for step in range(steps):

    # Choose a random spin
    i = np.random.randint(0, N)

    # Nearest neighbours
    L = spins[(i - 1) % N]
    R = spins[(i + 1) % N]

    # Change in energy
    dE = 2 * spins[i] * (J * (L + R) + H)

    # Metropolis algorithm
    if dE <= 0:
        spins[i] *= -1

    else:
        r = np.random.random()

        if r < np.exp(-dE / (kb * T)):
            spins[i] *= -1

    # Total magnetization
    M = np.sum(spins)
    M_val.append(M)

M_val = np.array(M_val)


# Calculations

M_mean = np.mean(M_val)
M_mean_sq = np.mean(M_val**2)

# Magnetic susceptibility
Chi = (M_mean_sq - M_mean**2) / (kb * T)


print("Mean Magnetization =", M_mean)
print("Mean Magnetization Squared =", M_mean_sq)
print("Magnetic Susceptibility =", Chi)
```
